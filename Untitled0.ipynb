```python
import time
import torch

context = """
Customer Rahul Patil has order #AB12345.
The order was shipped on 28 October 2026 and is currently out for delivery.
The delivery address is Hinjewadi, Pune.
Rahul's registered phone number is 9876543210.
Rahul's registered email is rahul.patil@gmail.com.
The order amount is ₹2,499 and payment status is successful.
Customers can cancel an order only before it is shipped.
"""

query = """
Hello, my name is Rahul. Where is my order #AB12345?
Was my payment successful?
Can I cancel the order?
"""

prompt = f"""
You are a customer support assistant.

Answer the customer query using ONLY the provided context.

Rules:
- Do not invent information.
- Do not use outside knowledge.
- If the context does not contain the answer, say "Information not available."
- Be concise.
- Answer all questions.

CONTEXT:
{context}

CUSTOMER QUERY:
{query}

ANSWER:
"""

messages = [
    {"role": "user", "content": prompt}
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = tokenizer(
    text,
    return_tensors="pt"
).to("cuda")

torch.cuda.synchronize()
start = time.perf_counter()

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=150,
        temperature=0.0,
        do_sample=False
    )

torch.cuda.synchronize()
latency = time.perf_counter() - start

new_tokens = outputs[0][inputs["input_ids"].shape[1]:]
answer = tokenizer.decode(
    new_tokens,
    skip_special_tokens=True
).strip()

print("=" * 50)
print("QWEN3-0.6B MAIN MODEL — RAG TEST")
print("=" * 50)

print("\nANSWER:")
print(answer)

print("\nLATENCY:")
print(f"{latency:.3f} seconds")
```

```python
import time
import torch
import json
import re

query = """
hello my nam is Rahul i want to know where my ordar AB12345 is
and can i cancle it? also plese tell me if my paymant was sucessful
"""

prompt = f"""
You are an NLP engine for a customer support chatbot.

TASK 1: SPELLING AND GRAMMAR CORRECTION
Correct the customer's query while preserving its meaning.

TASK 2: ENTITY EXTRACTION
Extract entities ONLY from the ORIGINAL CUSTOMER QUERY.

Important rules:
- Do NOT extract entities from the instructions.
- Do NOT invent entities.
- Do NOT use outside knowledge.
- Preserve entity values exactly as they appear in the ORIGINAL query.
- If an entity is not present, return null.
- "Rahul" is a name.
- "AB12345" is an order ID.
- Do not extract entities from the corrected query if they were not present in the original query.

Entity fields:
- name
- order_id
- phone
- email
- area
- city

Return ONLY valid JSON.
No markdown.
No explanation.
No <think> section.

Required format:

{{
  "corrected_query": "...",
  "entities": {{
    "name": null,
    "order_id": null,
    "phone": null,
    "email": null,
    "area": null,
    "city": null
  }}
}}

ORIGINAL CUSTOMER QUERY:
{query}

JSON:
"""

messages = [
    {"role": "user", "content": prompt}
]

# Disable thinking if supported by Qwen3 chat template
text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
    enable_thinking=False
)

inputs = tokenizer(text, return_tensors="pt").to("cuda")

torch.cuda.synchronize()
start = time.perf_counter()

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=250,
        do_sample=False
    )

torch.cuda.synchronize()
latency = time.perf_counter() - start

# Only generated tokens
new_tokens = outputs[0][inputs["input_ids"].shape[1]:]

raw_output = tokenizer.decode(
    new_tokens,
    skip_special_tokens=True
).strip()

print("========== RAW OUTPUT ==========")
print(raw_output)

print("\n========== LATENCY ==========")
print(f"{latency:.3f} seconds")

# Try to extract JSON
match = re.search(r'\{.*\}', raw_output, re.DOTALL)

if match:
    json_text = match.group(0)

    try:
        result = json.loads(json_text)

        print("\n========== PARSED JSON ==========")
        print(json.dumps(result, indent=2, ensure_ascii=False))

        print("\n========== CHECK ==========")

        expected_name = "Rahul"
        expected_order_id = "AB12345"

        entities = result.get("entities", {})

        print("Name:",
              entities.get("name") == expected_name)

        print("Order ID:",
              entities.get("order_id") == expected_order_id)

        print("Phone:",
              entities.get("phone") is None)

        print("Email:",
              entities.get("email") is None)

        print("Area:",
              entities.get("area") is None)

        print("City:",
              entities.get("city") is None)

        print("JSON VALID: True")

    except json.JSONDecodeError:
        print("\nJSON VALID: False")
else:
    print("\nJSON VALID: False")
```
